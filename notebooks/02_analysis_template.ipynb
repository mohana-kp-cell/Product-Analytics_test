## 4. Visualizations

Best Practice: Clear titles, labels, and legends.

#### Product-wise Revenue contribution
 Why? -> To identify which products drive your business revenue

```python
# Revenue calculation
df['revenue'] = df['quantity'] * df['price']

# Group revenue by product
product_rev = df.groupby('product_name')['revenue'].sum().sort_values(ascending=False)

print(product_rev)

# Plot
product_rev.head(10).plot(kind='bar', figsize=(8,4))
plt.title("Top Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```

#### Revenue vs Quantity

Why? -> 
To understand whether revenue comes from high price or high volume products.

```python
df[['quantity','revenue']].plot(kind='scatter', x='quantity', y='revenue')
plt.title("Quantity vs Revenue")
plt.show()
```

#### Revenue over time
Why? -> To detect seasonality, demand spikes, or declining periods

```python
df['date'] = pd.to_datetime(df['order_date'])
daily = df.groupby('date')['revenue'].sum()

daily.plot(figsize=(10,4))
plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

#### Category-wise revenue
Why? -> Shows which segment/category gives more business

```python
category_rev = df.groupby('category')['revenue'].sum().sort_values(ascending=False)
print(category_rev)

category_rev.plot(kind='bar', figsize=(8,4))
plt.title("Revenue by Category")
plt.show()
```

#### Top customers by revenue
Why? -> To identify key accounts

```python
top_customers = df.groupby('customer_id')['revenue'].sum().sort_values(ascending=False)
print(top_customers.head(10))
```
