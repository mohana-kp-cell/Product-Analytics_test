## 2. Load Data

```python
# Load from core module
import sys
sys.path.append('..')

# Or use relative import
df = pd.read_csv('../data/sample_sales.csv')
print(f"Data loaded: {len(df)} rows")
```

## 3. Analysis

Describe what this analysis does and why.

```python
# -------------------------------
# Customer Analysis
# -------------------------------

# 1. Calculate customer-level metrics
df = df.copy()
df['order_value'] = df['quantity'] * df['price']

customer_summary = (
    df.groupby('customer_id')
    .agg(
        total_spend=('order_value', 'sum'),
        total_orders=('order_id', 'nunique'),
        avg_quantity=('quantity', 'mean')
    )
    .reset_index()
)

customer_summary.head()
```

What This Analysis Does & Why

This analysis prepares a customer-level dataset to understand customer behavior. It includes:

1. total_spend

Measures how much each customer has spent.

Useful for identifying high-value customers.

2. total_orders

Counts how many orders each customer made.

Helps identify engaged or repeat customers.

3. avg_quantity

Average quantity purchased per order.

Shows buying pattern—customers who place large-quantity orders may represent wholesale or bulk buyers.

Why this analysis matters

Helps prioritize marketing and retention efforts.

Distinguishes between high-frequency vs. high-value customers.

Provides insights for segmentation, discount strategy, and personalization.

## 4. Visualizations

Best Practice: Clear titles, labels, and legends.

```python
# -------------------------------
# Customer Visualizations
# -------------------------------

import matplotlib.pyplot as plt

# ---- 1. Distribution of Customer Spend ----
plt.figure(figsize=(10, 6))
plt.hist(customer_summary['total_spend'], bins=30)
plt.title('Distribution of Total Customer Spend')
plt.xlabel('Total Spend')
plt.ylabel('Number of Customers')
plt.show()

# ---- 2. Top 10 Customers by Spend ----
top10 = customer_summary.sort_values(by='total_spend', ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(top10['customer_id'].astype(str), top10['total_spend'])
plt.title('Top 10 Customers by Total Spend')
plt.xlabel('Customer ID')
plt.ylabel('Total Spend')
plt.xticks(rotation=45)
plt.show()

# ---- 3. Relationship: Total Spend vs Number of Orders ----
plt.figure(figsize=(10, 6))
plt.scatter(customer_summary['total_orders'], customer_summary['total_spend'])
plt.title('Total Spend vs Number of Orders')
plt.xlabel('Number of Orders')
plt.ylabel('Total Spend')
plt.show()

# ---- 4. Distribution of Average Quantity per Customer ----
plt.figure(figsize=(10, 6))
plt.hist(customer_summary['avg_quantity'], bins=25)
plt.title('Distribution of Average Quantity per Customer')
plt.xlabel('Average Quantity')
plt.ylabel('Number of Customers')
plt.show()
```

## 5. Key Insights

**Key Insight #1** — Customer spend is highly skewed

A small number of customers contribute a large share of total revenue.
The top 10 customers stand out significantly in total spend, showing that the business depends heavily on a few high-value customers.

**Key Insight #2** — Many customers place only 1–2 orders

The distribution of total_orders shows most customers are low-frequency buyers.
Only a small group shows consistent repeat purchases.

**Key Insight #3** — Buying patterns vary widely

Customer avg_quantity distribution indicates:

Some customers usually buy in small units, suggesting casual/retail buyers.

Others consistently buy in large quantities, likely bulk/wholesale customers.

This split hints at multiple distinct customer segments.

**Recommendations**

Top spenders

This will protect and grow the most valuable segment.

Action Item 2 — Engage low-frequency customers

Create reactivation strategies such as:
