## Data Loading

````python
def load_sales_data(filepath='../data/sample_sales.csv'):
    """Load sales data from CSV.
    
    Best Practice: Use relative paths for repo data,
    mount points for production data.
    
    Args:
        filepath: Path to data file
        
    Returns:
        DataFrame with sales data
    """
    df = pd.read_csv(filepath)
    return df
````

## Summary Metrics Function

⚠️ **CONFLICT ZONE** ⚠️

This function will be modified by multiple team members in parallel.
- User A will add: Revenue metrics
- User B will add: Customer metrics

This simulates real team collaboration!

````python
def calculate_summary_metrics(df):
    """Calculate summary metrics from sales data.
    
    TODO: This function needs to be enhanced!
    
    User A: Add revenue metrics
    - total_revenue
    - average_order_value
    - revenue_by_category
    
    User B: Add customer metrics
    - unique_customers
    - average_quantity_per_order
    - top_customers (top 5 by spend)
    
    Best Practice: Document parameters and return values!
    
    Args:
        df: Sales DataFrame
        
    Returns:
        dict: Summary metrics
    """
    

    # -------- Customer Metrics (User B) --------

    # Unique customers
    unique_customers = df['customer_id'].nunique()

    # Average quantity per order
    average_quantity_per_order = df['quantity'].mean()

    # Compute order value per row
    df = df.copy()  # avoid modifying original dataframe
    df['order_value'] = df['quantity'] * df['price']

    # Top 5 customers by total spend
    top_customers = (
        df.groupby('customer_id')['order_value']
        .sum()
        .sort_values(ascending=False)
        .head(5)
        .reset_index()
        .rename(columns={'order_value': 'total_spend'})
        .to_dict('records')
    )

    # -------- Return Summary --------
    summary = {
        'total_orders': len(df),
        'unique_customers': unique_customers,
        'average_quantity_per_order': average_quantity_per_order,
        'top_customers': top_customers,
    }

    return summary
````

## Test the Function

Best Practice: Always test your functions!

````python
# Test
metrics = calculate_summary_metrics(df)
print("\nSummary Metrics:")
for key, value in metrics.items():
    print(f"  {key}: {value}")
````
